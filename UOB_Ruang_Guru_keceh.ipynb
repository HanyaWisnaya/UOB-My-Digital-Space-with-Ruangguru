```python
saldo = 0

print("Selamat datang di Bank Toot!\n")
bonusClaimed = 0

def cekSaldo():
  if saldo == 0:
    print("Saldo anda habis yahaha")
  else:
    print(f"Saldo anda: Rp.{saldo}")

def showMenu():
  print("[1] Cek Saldo")
  print("[2] Topup Saldo")
  print("[3] Tarik Tunai")
  if bonusClaimed == 0:
    print("[67] BONUS TERBATAS!! PILIH SEGERA!!")
  print("[0] Keluar")

def tarikSaldo(amount):
  if amount > saldo:
    print("Jumlah penarikan lebih besar dari saldomu.")
  return saldo - amount

def topupSaldo():
  try:
    amount = int(input("Masukkan nominal: Rp."))
    print(f"Topup berhasil! Saldomu sekarng: Rp.{saldo+amount}")
    return amount
  except:
    print("Nominal tidak valid!")
    return 0


while True:
  showMenu()

  try:
    option = int(input("\nPilih opsi: "))
  except:
    print("Opsi tidak valid.")
  print("")

  if option == 0:
    print("Terimakasih telah menggunakan Bank Toot!")
    break
  elif option == 1:
    cekSaldo()
  elif option == 2:
    saldo += topupSaldo()
  elif option == 3:
    amount = input("Masukkan jumlah penarikan: Rp.")
    tarikSaldo(amount)
    print(f"Sukses! Saldo anda sekarang: Rp.{saldo}")
  elif option == 67:
    saldo = -6767676767
    bonusClaimed = 1
    print("Sukses mengkalim bonus! Cek saldo anda dengan opsi 1.")
  print("")
```

```python
riwayatPengeluaran = []
profilAplikasi = {
    "username": "Demonstrasi",
    "currency": "IDR"
}

def addPurchase():
  print("\n+=== TAMBAH PENGELUARAN ===+")
  nama = input("Masukkan nama pengeluaran: ").strip()

  try:
    nominal = float(input("Masukkan nominal: Rp."))
    transaksi = {
        "nama": nama,
        "nominal": nominal
    }
    riwayatPengeluaran.append(transaksi)
    print(f"[+] Sukses mencatat: {nama} sebesar Rp.{nominal:,.0f}!")
  except ValueError:
    print("[-] Nominal yang dimasukkan harus berupa angka!")

def seeHistory():
  print("\n+=== RIWAYAT PENGELUARAN ===+")
  if not riwayatPengeluaran:
    print("[-] Tidak ada riwayat pengeluaran!")
  else:
    total = 0
    print("No. | \tNama Pengeluaran | \tNominal")
    print("------------------------------------------")
    for id, item in enumerate(riwayatPengeluaran, 1):
      print(f"{id}. | \t{item['nama']:<22} | \tRp.{item['nominal']:,.0f}")
      total += item['nominal']

    print("------------------------------------------")
    print(f"Total pengeluaran: Rp.{total:,.0f}")

def deletePurchase():
  print("\n+=== HAPUS TRANSAKSI TERAKHIR ===+")

  if not riwayatPengeluaran:
    print("[-] Tidak ada riwayat pengeluaran!")
  else:
    deleted = riwayatPengeluaran.pop()
    print(f"[+] Berhasil menghapus transaksi: {deleted['nama']} (Rp.{deleted['nominal']:,.0f})")

print(f"=== APLIKASI KEUANGAN [{profilAplikasi['username']}] ===")

while True:
    print("\n+=== MENU UTAMA ===+")
    print("1. Tambah Pengeluaran")
    print("2. Lihat Riwayat & Total")
    print("3. Hapus Transaksi Terakhir")
    print("4. Keluar")

    pilihan = input("Pilih menu (1-4): ").strip()

    if pilihan == "1":
        addPurchase()
    elif pilihan == "2":
        seeHistory()
    elif pilihan == "3":
        deletePurchase()
    elif pilihan == "4":
        print("[-] Program selesai. Data tersimpan sementara di memori runtime.")
        break
    else:
        print("[-] Pilihan tidak valid. Silakan pilih 1-4.")
```
